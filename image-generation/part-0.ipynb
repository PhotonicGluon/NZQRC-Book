### Chapter 3: Relations

The "is born in" relation diagram.

```python
%%handle_display
%%manim --output_file 0-relations-is-born-in.png RelationDiagram

# TODO: Continue working on this

class RelationDiagram(Scene):
    def construct(self):
        # Create the ellipses
        ellipse_a = Ellipse(width=2.5, height=4, color=ORANGE)
        ellipse_a.set_fill(ORANGE, opacity=0.25)
        ellipse_b = Ellipse(width=2.5, height=4, color=BLUE)
        ellipse_a.set_fill(BLUE, opacity=0.25)
        
        # Position the ellipses
        ellipse_a.shift(LEFT * 3)
        ellipse_b.shift(RIGHT * 3)
        
        # Create labels for the ellipses
        label_a = Text("A", color=BLACK).next_to(ellipse_a, UP)
        label_b = Text("B", color=BLACK).next_to(ellipse_b, UP)
        
        # Create elements for set A
        elements_a = VGroup(
            Text("1", color=BLACK),
            Text("2", color=BLACK),
            Text("5", color=BLACK),
            Text("7", color=BLACK)
        ).arrange(DOWN, buff=0.5).move_to(ellipse_a)
        
        # Create elements for set B
        elements_b = VGroup(
            Text("a", color=BLACK),
            Text("c", color=BLACK),
            Text("m", color=BLACK),
            Text("n", color=BLACK)
        ).arrange(DOWN, buff=0.5).move_to(ellipse_b)
        
        # Create arrows
        arrows = VGroup(
            Arrow(elements_a[0], elements_b[0], color=RED),
            Arrow(elements_a[1], elements_b[1], color=RED),
            Arrow(elements_a[2], elements_b[1], color=RED),
            Arrow(elements_a[3], elements_b[3], color=RED)
        )
        
        # Add all elements to the scene
        self.add(ellipse_a, ellipse_b, label_a, label_b, elements_a, elements_b, arrows)
```
