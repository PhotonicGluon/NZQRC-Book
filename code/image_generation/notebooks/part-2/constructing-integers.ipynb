### Natural Numbers on a Number Line

Just the plain ol' number line.

```python
%%generate_image number-line_raw NumberLineDiagram
    
FONT_SIZE = 48


class NumberLineDiagram(Scene):
    def construct(self):
        # Create the line
        line = NumberLine(
            x_range=[0, 9.99, 1],
            numbers_to_exclude=[10],
            color=BLACK,
            numbers_with_elongated_ticks=[0],
            include_numbers=True,
            label_constructor=Text,
            font_size=FONT_SIZE,
            line_to_number_buff=0.35
        )
        for n in line.numbers:
            n.set_color(BLACK)
        line.add_tip(tip_shape=StealthTip, tip_length=0.2)
        
        self.add(line)
```

First 3 then add 4.

```python
%%generate_image number-line_add-3-then-4 NumberLineDiagram
    
FONT_SIZE = 48


class NumberLineDiagram(Scene):
    def construct(self):
        # Create the line
        line = NumberLine(
            x_range=[0, 9.99, 1],
            numbers_to_exclude=[10],
            color=BLACK,
            numbers_with_elongated_ticks=[0],
            include_numbers=True,
            label_constructor=Text,
            font_size=FONT_SIZE,
            line_to_number_buff=0.35
        )
        for n in line.numbers:
            n.set_color(BLACK)
        line.add_tip(tip_shape=StealthTip, tip_length=0.2)

        self.add(line)
        
        # Add an arrow for 3
        three_arrow = Arrow(line.n2p(0), line.n2p(3), color=RED, buff=0)
        three_arrow.shift(0.5*UP)
        self.add(three_arrow)

        # Add an arrow for 4
        four_arrow = Arrow(line.n2p(3), line.n2p(7), color=BLUE, buff=0)
        four_arrow.shift(0.5*UP)
        self.add(four_arrow)
```

```python
%%generate_image number-line_add-4-then-3 NumberLineDiagram
    
FONT_SIZE = 48


class NumberLineDiagram(Scene):
    def construct(self):
        # Create the line
        line = NumberLine(
            x_range=[0, 9.99, 1],
            numbers_to_exclude=[10],
            color=BLACK,
            numbers_with_elongated_ticks=[0],
            include_numbers=True,
            label_constructor=Text,
            font_size=FONT_SIZE,
            line_to_number_buff=0.35
        )
        for n in line.numbers:
            n.set_color(BLACK)
        line.add_tip(tip_shape=StealthTip, tip_length=0.2)

        self.add(line)

        # Add an arrow for 4
        four_arrow = Arrow(line.n2p(0), line.n2p(4), color=RED, buff=0)
        four_arrow.shift(0.5*UP)
        self.add(four_arrow)

        # Add an arrow for 3
        three_arrow = Arrow(line.n2p(4), line.n2p(7), color=BLUE, buff=0)
        three_arrow.shift(0.5*UP)
        self.add(three_arrow)
```

A leftward 3 arrow from 7.

```python
%%generate_image number-line_sub-3-from-7 NumberLineDiagram
    
FONT_SIZE = 48


class NumberLineDiagram(Scene):
    def construct(self):
        # Create the line
        line = NumberLine(
            x_range=[0, 9.99, 1],
            numbers_to_exclude=[10],
            color=BLACK,
            numbers_with_elongated_ticks=[0],
            include_numbers=True,
            label_constructor=Text,
            font_size=FONT_SIZE,
            line_to_number_buff=0.35
        )
        for n in line.numbers:
            n.set_color(BLACK)
        line.add_tip(tip_shape=StealthTip, tip_length=0.2)

        self.add(line)

        # Add an arrow for 7
        seven_arrow = Arrow(line.n2p(0), line.n2p(7), color=RED, buff=0)
        seven_arrow.shift(0.5*UP)
        self.add(seven_arrow)

        # Add an arrow for 3
        three_arrow = Arrow(line.n2p(7), line.n2p(4), color=BLUE, buff=0)
        three_arrow.shift(0.85*UP)
        self.add(three_arrow)
```

A leftward 3 arrow from 1.

```python
%%generate_image number-line_sub-3-from-1 NumberLineDiagram
    
FONT_SIZE = 48


class NumberLineDiagram(Scene):
    def construct(self):
        # Create the line
        line = NumberLine(
            x_range=[0, 4.99, 1],
            numbers_to_exclude=[5],
            color=BLACK,
            numbers_with_elongated_ticks=[0],
            include_numbers=True,
            label_constructor=Text,
            font_size=FONT_SIZE,
            line_to_number_buff=0.35
        )
        for n in line.numbers:
            n.set_color(BLACK)
        line.add_tip(tip_shape=StealthTip, tip_length=0.2)

        self.add(line)

        # Add an arrow for 1
        seven_arrow = Arrow(line.n2p(0), line.n2p(1), color=RED, buff=0)
        seven_arrow.shift(0.5*UP)
        self.add(seven_arrow)

        # Add an arrow for "-2"
        three_arrow = Arrow(line.n2p(1), line.n2p(-2), color=BLUE, buff=0)
        three_arrow.shift(0.85*UP)
        self.add(three_arrow)

        # Add a question mark for the resulting "-2"
        question_mark = Text("?", color=BLACK, font_size=FONT_SIZE*1.5)
        question_mark.move_to(line.n2p(-2))
        self.add(question_mark)
```
