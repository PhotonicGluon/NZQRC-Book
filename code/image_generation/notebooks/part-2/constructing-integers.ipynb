### "Leftward" and "Rightward" Shifts

Helper function to create the shift arrows.

```python
def create_shift_arrows(scene: Scene, line: NumberLine, right: int, left: int) -> tuple[Arrow, Arrow]:
    # Add an arrow for `right`
    right_arrow = Arrow(line.n2p(0), line.n2p(right), color=RED, buff=0)
    right_arrow.shift(0.5*UP)
    scene.add(right_arrow)

    right_arrow.add(
        Text(
            f"Right by {right}", color=BLACK, font_size=0.75*FONT_SIZE
        ).next_to(right_arrow, RIGHT).shift(0.025*DOWN)
    )

    # Add an arrow for `left`
    left_arrow = Arrow(line.n2p(right), line.n2p(right-left), color=BLUE, buff=0)
    left_arrow.shift(0.85*UP)
    scene.add(left_arrow)

    left_arrow.add(
        Text(
            f"Left by {left}", color=BLACK, font_size=0.75*FONT_SIZE
        ).next_to(left_arrow, LEFT).shift(0.025*DOWN)
    )

    return right_arrow, left_arrow
```

Other shifts that result in 4.

```python
%%generate_image lr-shifts_4-shifts-2 NumberLineDiagram
    
FONT_SIZE = 32

SHIFTS = [
    (5, 1),
    (6, 2),
    (7, 3),
    (8, 4),
]
NUM_SHIFTS = len(SHIFTS)
BUFF = 1.75

class NumberLineDiagram(Scene):
    def construct(self):        
        for i, (right, left) in enumerate(SHIFTS):
            # Create the line
            line = NumberLine(
                x_range=[0, 9.99, 1],
                numbers_to_include=[0],
                color=BLACK,
                numbers_with_elongated_ticks=[0],
                include_numbers=False,
                label_constructor=Text,
                font_size=FONT_SIZE,
                line_to_number_buff=0.35
            )
            for n in line.numbers:
                n.set_color(BLACK)
            line.add_tip(tip_shape=StealthTip, tip_length=0.2)
            
            self.add(line)
    
            # Create shift
            right_arrow, left_arrow = create_shift_arrows(self, line, right, left)

            # Create a group for all the things, and adjust
            group = Group(line, right_arrow, left_arrow)

            group.shift((i - (NUM_SHIFTS-1)/2) * BUFF * DOWN)
```

### Defining Integers

Create a grid of pairwise naturals.

```python
%%generate_image integers_number-grid GridDiagram

FONT_SIZE = 32
RANGE = (0, 5.5, 1)
ROUNDED_RANGE = (int(np.ceil(RANGE[0])), int(np.ceil(RANGE[1])), RANGE[2])


class GridDiagram(Scene):
    def construct(self):
        # Define the number plane
        plane = NumberPlane(
            x_range=RANGE,
            y_range=RANGE,
            background_line_style={
                "stroke_color": GRAY_C,
                "stroke_width": 2
            },
            axis_config={
                "include_numbers": True,
                "color": BLACK,
                "stroke_width": 4,
                "label_constructor": Text,
                "font_size": FONT_SIZE
            }
        )
        for n in plane.x_axis.numbers:
            n.set_color(BLACK)
        for n in plane.y_axis.numbers:
            n.set_color(BLACK)
            flip_about_line(n, plane.y_axis, handle_orientation=False)
            
        self.add(plane)

        # Add missing 0
        zero = Text("0", font_size=FONT_SIZE, color=BLACK)
        zero.move_to(plane.coords_to_point(0, 0))
        zero.shift(0.2*(DOWN+LEFT))
        self.add(zero)

        # Draw lines representing the equivalence classes
        for k in range(-ROUNDED_RANGE[1] + 1, ROUNDED_RANGE[1]):
            if k < 0:
                min_x = RANGE[0]
                min_y = RANGE[0] - k

                max_x = RANGE[1] + k
                max_y = RANGE[1]

            else:
                min_x = RANGE[0] + k
                min_y = RANGE[0]

                max_x = RANGE[1]
                max_y = RANGE[1] - k

            line = DashedLine(
                plane.coords_to_point(min_x, min_y),
                plane.coords_to_point(max_x, max_y),
                color=BLACK,
                dashed_ratio=0.35
            )
            self.add(line)

            integer = Text(str(k), font_size=FONT_SIZE, color=BLUE_E)
            integer.move_to(line.get_end() + 0.35*unit_vector(line.get_vector()))
            self.add(integer)
            
        # Create intersection points
        for x in range(ROUNDED_RANGE[0], ROUNDED_RANGE[1], ROUNDED_RANGE[2]):
            for y in range(ROUNDED_RANGE[0], ROUNDED_RANGE[1], ROUNDED_RANGE[2]):
                dot = Dot(plane.coords_to_point(x, y), color=RED)
                self.add(dot)
```

### Ordering of Integers

```python
%%generate_image integer-ordering HasseDiagram


class HasseDiagram(Scene):
    def construct(self):
        # Create vertices and edges
        vertices = [r"\vdots", "-2", "-1", "0", "1", "2", r"\vdots "]  # Need a space on the last one to make vertices unique
        edges = [(vertices[i], vertices[i+1]) for i in range(len(vertices)-1)]
        
        # Create the Hasse graph
        graph = HasseGraph(
            vertices,
            edges,
            label_colour=BLACK,
            perform_correction=False,
            label_buff=0.3,
            edge_config={
                "stroke_color": BLACK
            },
            layout_config={
                "scale_x": 2,
                "scale_y": 0.5,
            }
        )
        # print(list(graph.edges.keys()))
        
        self.add(graph)
```
